**A. Membangun Katalog (Create)** *(bobot 20%)*

Buat database baru bernama `kopi_nusantara`, collection bernama `menu`. Sisipkan **minimal 12 dokumen** menu dengan field bervariasi sesuai kategorinya:
- Menu kategori **"Kopi"**: wajib memiliki field `nama_menu`, `harga`, `kategori`, dan `tingkat_kepahitan` (skala 1-5).
- Menu kategori **"Makanan"**: wajib memiliki field `nama_menu`, `harga`, `kategori`, dan `pedas_level` (skala 0-3).
- Boleh menambahkan field lain sesuai kreativitas anda (mis. `stok`, `rekomendasi_barista`).

**B. Kueri Bertingkat (Read)** *(bobot 20%)*

Tampilkan hasil dari **tiga** kueri berikut :
1. Seluruh menu kategori "Kopi" dengan `tingkat_kepahitan` di atas 3.
2. Seluruh menu dengan `harga` di antara 15000 dan 30000.
3. Menu yang **tidak memiliki** field `pedas_level` sama sekali.

```python
# Kueri 1: Seluruh menu kategori "Kopi" dengan tingkat_kepahitan > 3
print("=== Kueri 1: Menu Kopi dengan Tingkat Kepahitan > 3 ===")
kueri_1 = koleksi_menu.find({"kategori": "Kopi", "tingkat_kepahitan": {"$gt": 3}})
for m in kueri_1:
    print(f"- {m['nama_menu']} (Kepahitan: {m['tingkat_kepahitan']}/5) | Rp{m['harga']:,}")

# Kueri 2: Seluruh menu dengan harga di antara 15000 dan 30000
print("\n")
print("=== Kueri 2: Menu dengan Harga Rp15.000 - Rp30.000 ===")
kueri_2 = koleksi_menu.find({"harga": {"$gte": 15000, "$lte": 30000}})
for m in kueri_2:
    print(f"- {m['nama_menu']} [{m['kategori']}] | Rp{m['harga']:,}")

# Kueri 3: Menu yang TIDAK memiliki field 'pedas_level' sama sekali
print("\n")
print("=== Kueri 3: Menu yang Tidak Memiliki Field 'pedas_level' ===")
kueri_3 = koleksi_menu.find({"pedas_level": {"$exists": False}})
for m in kueri_3:
    print(f"- {m['nama_menu']} [{m['kategori']}] | Rp{m['harga']:,}")
```

**C. Pembaruan Data (Update)** *(bobot 15%)*

Tambahkan field `status` bernilai `"Tersedia"` ke seluruh dokumen menggunakan `update_many()`. Kemudian, ubah `status` menjadi `"Habis"` khusus untuk 2 menu pilihan anda menggunakan `update_one()`.

```python
# 1. Tambahkan field status: "Tersedia" ke seluruh dokumen
koleksi_menu.update_many({}, {"$set": {"status": "Tersedia"}})

# 2. Ubah status menjadi "Habis" khusus untuk 2 menu pilihan
koleksi_menu.update_one({"nama_menu": "Kopi Tubruk Kintamani"}, {"$set": {"status": "Habis"}})
koleksi_menu.update_one({"nama_menu": "Ayam Geprek Sambal Korek"}, {"$set": {"status": "Habis"}})

# Verifikasi hasil update
print("=== Daftar Menu Setelah Diperbarui ===")
for m in koleksi_menu.find({}, {"_id": 0, "nama_menu": 1, "kategori": 1, "status": 1}):
    print(f"- {m['nama_menu']} [{m['kategori']}] -> Status: {m['status']}")
```

**D. Integrasi ke Spark** *(bobot 25%)*

Ambil seluruh data dari collection `menu`, ubah menjadi Spark DataFrame (ikuti pola Sub-bab 9.6 — perhatikan field yang mungkin kosong/`null` pada sebagian dokumen, seperti `tingkat_kepahitan` yang hanya dimiliki menu Kopi). Tampilkan:
1. `printSchema()` hasil konversi.
2. Rata-rata `harga` per `kategori` menggunakan `groupBy`.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg, count
import pandas as pd

# Inisialisasi SparkSession
spark = SparkSession.builder \
    .appName("Tugas9-KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

# 1. Ambil data dari MongoDB
data_mongo = list(koleksi_menu.find({}))

# 2. Buat Pandas DataFrame
df_pandas = pd.DataFrame(data_mongo)

# 3. PERBAIKAN: Konversi kolom _id dan bersihkan nilai NaN/None agar PySpark tidak bingung
df_pandas["_id"] = df_pandas["_id"].astype(str)

# Ubah semua nilai NaN menjadi None agar terbaca sebagai null di PySpark tanpa bentrok tipe data
df_pandas = df_pandas.where(pd.notnull(df_pandas), None)

# Pastikan kolom opsional yang tipe datanya campuran diubah menjadi string / format aman
for col in df_pandas.columns:
    if col not in ["harga", "tingkat_kepahitan", "pedas_level", "stok"]:
        df_pandas[col] = df_pandas[col].astype(str)

# 4. Buat Spark DataFrame
df_spark_menu = spark.createDataFrame(df_pandas)

# D.1 Tampilkan Schema dan Data
print("=== Schema Spark DataFrame ===")
df_spark_menu.printSchema()

print("=== Data Spark DataFrame ===")
df_spark_menu.show(truncate=False)

# D.2 Rata-rata Harga per Kategori
print("=== Rata-rata Harga per Kategori ===")
df_spark_menu.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga"),
    count("nama_menu").alias("jumlah_menu")
).show()
```

**E. Kesimpulan** *(bobot 20%)*

Tulis pada markdown cell (**minimal 100 kata**): apa tantangan yang anda temui saat mengonversi data MongoDB (dengan field yang tidak seragam antar dokumen) menjadi Spark DataFrame (yang mengharuskan skema tabel seragam)? Bagaimana anda mengatasinya?

*Tantangan utama yang saya hadapi saat mengonversi data MongoDB (yang memiliki flexible schema) ke dalam Spark DataFrame adalah perbedaan fundamental dalam penanganan struktur data. Di MongoDB, dokumen sangat fleksibel; sebagai contoh, menu kategori "Kopi" memiliki field tingkat_kepahitan, sementara menu "Makanan" memiliki pedas_level. Ketika data ini ditarik dan dikonversi menjadi Spark DataFrame, Spark menuntut adanya skema tabel yang seragam (setiap baris harus memiliki kolom yang sama). Akibatnya, field yang tidak dimiliki oleh suatu dokumen akan diisi dengan nilai null (kosong), dan jika tidak ditangani dengan benar, hal ini dapat menyebabkan error saat Spark mencoba menginferensi tipe data (misalnya, Spark bingung menentukan tipe data untuk kolom yang seluruhnya kosong).*

*Untuk mengatasinya, saya memanfaatkan pandas DataFrame sebagai jembatan (perantara). Saya memastikan tipe data khusus MongoDB seperti ObjectId diubah menjadi string, dan menangani kolom yang berisi list/array agar bisa dibaca Spark. Selain itu, dengan membiarkan pandas mengisi field yang tidak ada dengan NaN (yang kemudian dibaca Spark sebagai null), saya bisa memastikan konversi berjalan lancar. Jika inferensi skema Spark masih gagal, solusi alternatifnya adalah dengan mendefinisikan skema secara eksplisit menggunakan StructType di PySpark, atau mengisi nilai default pada data yang kosong sebelum dikonversi.*

### Eksplorasi: Analisis Operasional & Manajemen Inventaris Kafe

```python
from pyspark.sql.functions import col, desc

# 1. Filter Menu Rekomendasi Barista yang Masih Tersedia
print("=== Menu Rekomendasi Barista (Status: Tersedia) ===")
df_spark_menu.filter((col("rekomendasi_barista") == "True") & (col("status") == "Tersedia")) \
    .select("nama_menu", "kategori", "harga", "stok") \
    .show(truncate=False)

# 2. Urutan Top 5 Menu dengan Stok Terbanyak (Inventory Analysis)
print("=== Top 5 Menu dengan Stok Terbanyak ===")
df_spark_menu.select("nama_menu", "kategori", "stok", "status") \
    .orderBy(desc("stok")) \
    .limit(5) \
    .show(truncate=False)
```
